```python
customer = {"name": "Rahul","phone": "","email": "rahul@gmail.com","city": None}
missing = 0
for i in customer:
    if customer[i] == None or customer[i] == "":
        missing += 1
print("Missing value:",missing)
if missing == 0:
    print("Complete")
elif missing <= 2:
    print("Partially completed")
else:
    print("Incomplete")
    
```
