# Set


```python
language = ["Python", "Java","Python", "C++", "Java", "Python"]
print(list(set(language)))
```

```python
gymA = {"Ravi", "Sneha", "Kiran", "Alok"}
gymB = {"Kiran", "Meera", "Alok", "Priya"}
print(gymA | gymB)
```

```python
print(gymA & gymB)
```

```python
print(gymA - gymB)
```

```python
a = "priya.sharma@gmail.com"
b = {"gmail.com", "yahoo.com", "outlook.com"}
print("gmail.com" in b)
```

```python
a = ("Delhi", "Agra", "Jaipur", "Delhi", "Agra")
b = set(a)
print(b)
print(len(b))
```

```python
morning = ["Ana", "Ben", "Cara", "Ana"]
afternoon = ["Ben", "Dev", "Cara"]
print(set(morning) - set(afternoon))
```

```python
usernames = {"troll99", "hacker007","spammer1"}
print("hacker007" in usernames)
print("This user name is already banned")
```

```python
post1 = ["vegan", "quick", "dinner"]
post2 =["dinner", "vegan", "healthy"]
print(set(post1) & set(post2))
```

```python
a = ("Mumbai", "Delhi", "Pune","Mumbai", "Chennai")
print(set(a))
```

```python
issued = {"Dune", "1984", "Emma"}
print("1984" in issued )
```

```python
round1 = ["Alex", "Bella", "Chris", "Alex"]
round2 = ["Bella", "Dan", "Chris", "Eva"]
print(set(round1) ^ set(round2))
```
