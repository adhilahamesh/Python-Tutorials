# Exception Handling in Python — Practical Notebook

When Python hits something it can't do (dividing by zero, converting "abc" into a number, opening a file that isn't there), it stops the whole program with an error. Exception handling lets your program catch the problem, deal with it, and keep running. Real-world data is messy, so this is something you'll use constantly.

## `try` and `except`

Put the risky code in `try`. If it fails, Python jumps to `except` instead of crashing.

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("You can't divide by zero")

print("The program carries on")
```

```python
try:
    number = int("abc")
except ValueError:
    print("That is not a valid number")
```

## The errors you'll meet most often

Each error has a name that tells you what went wrong.

```python
student = {"name": "Asha", "age": 21}

try:
    print(student["city"])
except KeyError:
    print("KeyError: that key does not exist")
```

```python
scores = [70, 80, 90]

try:
    print(scores[5])
except IndexError:
    print("IndexError: that position does not exist")
```

```python
try:
    print("5" + 2)
except TypeError:
    print("TypeError: can't add text and a number")
```

```python
try:
    open("missing_file.txt")
except FileNotFoundError:
    print("FileNotFoundError: that file does not exist")
```

## Handling different errors differently

You can have more than one `except`, each with its own message.

```python
value = "0"

try:
    result = 100 / int(value)
    print(result)
except ValueError:
    print("Please give a number")
except ZeroDivisionError:
    print("The number can't be zero")
```

## Seeing the actual error message

`except Exception as e` catches any error and lets you print what Python said about it.

```python
try:
    print(int("12.5"))
except Exception as e:
    print("Something went wrong:", e)
```

## `else` and `finally`

`else` runs only if nothing went wrong. `finally` runs no matter what — useful for tidy-up messages.

```python
try:
    price = float("199.50")
except ValueError:
    print("Invalid price")
else:
    print("Price accepted:", price)
finally:
    print("Price check finished")
```

```python
try:
    price = float("free")
except ValueError:
    print("Invalid price")
else:
    print("Price accepted:", price)
finally:
    print("Price check finished")
```

## Raising your own error

Sometimes a value is technically allowed by Python but wrong for your program. `raise` lets you stop with your own message.

```python
def set_temperature(celsius):
    if celsius < -273:
        raise ValueError("Temperature can't be below absolute zero")
    return celsius

try:
    set_temperature(-500)
except ValueError as e:
    print(e)
```

```python
try:
    a = int(input("Enter a number"))
    if a<=0 or a>150:
        raise Exception ("Error")
    print(a)
except Exception as e:
    print(e)
```

```python
def age(a):
    if a<=0 or 150<a:
        raise Exception("invalid age")
    return a
try:
    print(age(0))
except Exception as e:
    print(e)
```

## A useful pattern: skipping bad values

Real datasets have bad entries. A loop with `try`/`except` lets you keep the good ones and move past the rest.

```python
raw = ["8", "15", "n/a", "22"]
clean = []

for item in raw:
    try:
        clean.append(int(item))
    except ValueError:
        print("Skipped:", item)

print(clean)
```

```python
while True:
    try:
        quantity = int(input("Enter quantity: "))
        break
    except ValueError:
        print("Numbers only, please")

print("You ordered", quantity)
```
