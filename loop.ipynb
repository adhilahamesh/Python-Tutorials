```python
import random
one=0
two=0 
three=0
four=0
five=0
six=0
for i in range(100):
    dice = random.randint(1, 6)
    if dice == 1:
        one +=1
    elif dice == 2:
        two +=1
    elif dice == 3:
       three +=1
    elif dice == 4:
        four +=1
    elif dice == 5:
        five +=1
    else:
        six +=1
print("1.",one)
print("2.",two)
print("3.",three)
print("4.",four)
print("5.",five)
print("6.",six)
```

```python
import random
num = random.randint(1,50)
attempts = 0

guess = 0

while guess != num:
    guess = int(input("Guess the number:"))
    attempts +=1

    if guess>num:
        print("Too High")
    elif guess<num:
        print("Too Low")
    else:
        print("Correct")
print("Attempts Taken",attempts)
```

```python
while True:
    password=input("Enter password")
    upper = False
    lower = False
    digit = False
    char = False
    for i in password:
        if i.isupper():
            upper = True
        elif i.islower():
            lower = True
        elif i.isdigit():
            digit = True
        else:
            char = True

    if len(password)>8 and upper and lower and digit and char:
        print("Valid password")
        break   
    else:
        print("Try again")
```
