```python
a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))

gcd = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

print("GCD =", gcd)
```

```python
pin = "1234"
attempts = 0
while attempts < 3:
    a = input("Enter pin")
    if a in pin:
        print("Logged in")
        break
    else:
        attempts+=1
        print("Wrong pin")
if attempts == 3:
   print("Locked")
```
