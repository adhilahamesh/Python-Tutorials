# Modules and Imports in Python — Practical Notebook

A module is a file of ready-made Python code (functions, values, tools) that you can bring into your own program instead of writing it from scratch. Python comes with many built-in modules, and every data science library you'll use later (NumPy, Pandas, Matplotlib) is a module too.

## Bringing in a whole module

`import` loads the module. You then use its contents with a dot: `module_name.thing`.

```python
import math

print(math.sqrt(81))
```

```python
print(math.pi)
```

## Giving a module a shorter name

`as` lets you choose a nickname, which saves typing. You'll see this constantly in data science (`import numpy as np`).

```python
import math as m

print(m.sqrt(49))
```

## Bringing in only what you need

`from ... import ...` loads just the names you ask for, and you use them without the module prefix.

```python
from math import ceil, floor

print(ceil(4.2))
print(floor(4.8))
```

## Finding out what a module contains

`dir()` lists everything inside a module, and `help()` explains one item.

```python
import math

print(dir(math))
```

```python
help(math.factorial)
```

## The `random` module

Used for picking or shuffling things at random. Setting a `seed` first makes the "random" results repeat every time, which is handy while learning.

```python
import random

random.seed(10)
print(random.randint(1, 6))
```

```python
colors = ["red", "green", "blue", "yellow"]
print(random.choice(colors))
```

```python
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
```

```python
names = ["Anu", "Bala", "Chitra", "Dev", "Esha"]
print(random.sample(names, 2))
```

```python
print(random.random())
```

## The `statistics` module

Quick summary numbers for a list of values — very handy when you start analysing data.

```python
import statistics

scores = [10, 20, 20, 30, 40]
print(statistics.mean(scores))
```

```python
print(statistics.median(scores))
```

```python
print(statistics.mode(scores))
```

## The `datetime` module

For working with dates and times.

```python
from datetime import date

launch = date(2026, 10, 12)
print(launch)
```

```python
print(launch.year, launch.month, launch.day)
```

```python
print(launch.strftime("%d %B %Y"))
```

```python
print(launch.strftime("%d/%m/%y"))
```

```python
from datetime import timedelta

print(launch + timedelta(days=7))
```

```python
deadline = date(2026, 12, 25)
gap = deadline - launch
print(gap.days)
```

```python
today = date.today()
print(type(today))
```
